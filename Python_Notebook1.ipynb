Purpose: Build a reusable function to detect outliers using the Inter-Quartile Range test

```python
# Define a function to detect outliers in any numeric variable using IQR method
def find_outliers_IQR(data):
    # Calculate the first quartile (25th percentile)
    q1 = data.quantile(0.25)
    # Calculate the third quartile (75th percentile)
    q3 = data.quantile(0.75)
    # Calculate the interquartile range
    IQR = q3 - q1
    # Identify values outside lower fence (Q1-1.5*IQR) and upper fence (Q3+1.5*IQR)
    outliers = data[((data < (q1 - 1.5*IQR)) | (data > (q3 + 1.5*IQR)))]
    return outliers
```

Purpose: Call the IQR function to detect outlier instances in the age variable

```python
# Call the IQR outlier detection function on the age variable
outliers = find_outliers_IQR(data['age'])
# Print the total count of detected outliers in age
print("number of outliers: " + str(len(outliers)))
# Display the outlier records
outliers
```
